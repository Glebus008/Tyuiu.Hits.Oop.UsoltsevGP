<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Notification в C#, который будет представлять уведомления
пользователям. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Notification:
• Атрибуты: ID уведомления (NotificationId), Текст уведомления (MessageText),
Тип уведомления (Type).
• Методы:
o DisplayNotification(): метод для отображения уведомления
пользователю.
o SendNotification(): метод для отправки уведомления.
o GetNotificationDetails(): метод для получения деталей уведомления.
Требования к производным классам:
1. EmailУведомление (EmailNotification): Должно содержать дополнительные
атрибуты, такие как Адрес электронной почты (EmailAddress).
Метод SendNotification() должен быть переопределен для отправки
уведомления по электронной почте.
2. SMSУведомление (SMSNotification): Должно содержать дополнительные
атрибуты, такие как Номер телефона (PhoneNumber).
Метод SendNotification() должен быть переопределен для отправки
уведомления через SMS.
3. PushУведомление (PushNotification) (если требуется третий класс): Должно
содержать дополнительные атрибуты, такие как Платформа (Platform,
например, iOS или Android). Метод DisplayNotification() должен быть
переопределен для отображения уведомления на мобильной платформе.

Дополнительное задание

Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями

```
using System;
using System.Collections.Generic;

public interface IArchivable
{
    bool IsArchived { get; set; }
    void Archive();
}

public class Notification : IArchivable
{
    private int notificationId;
    private string messageText;
    private string type;

    public DateTime Timestamp { get; set; } = DateTime.Now;
    public int Priority { get; set; } = 1;
    public bool IsSent { get; set; } = false;
    public string SenderName { get; set; } = "Система";
    public int RetryLimit { get; set; } = 3;

    public bool IsArchived { get; set; }

    public int NotificationId
    {
        get => notificationId;
        set => notificationId = value;
    }

    public string MessageText
    {
        get => messageText;
        set => messageText = value;
    }

    public string Type
    {
        get => type;
        set => type = value;
    }

    public Notification(int notificationId, string messageText, string type)
    {
        NotificationId = notificationId;
        MessageText = messageText;
        Type = type;
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine($"[Уведомление #{NotificationId}] От: {SenderName} | Тип: {Type} | Текст: \"{MessageText}\"");
    }

    public virtual void LogMetadata()
    {
        Console.WriteLine($"[Лог] ID: {NotificationId}, Приоритет: {Priority}, Время: {Timestamp}");
    }

    public virtual void Validate()
    {
        Console.WriteLine($"[Валидация] Проверка уведомления #{NotificationId} на корректность...");
    }

    public virtual void SendNotification()
    {
        IsSent = true;
        Console.WriteLine($"Отправка стандартного уведомления #{NotificationId} от {SenderName}...");
    }

    public virtual void SendNotification(int retryCount)
    {
        Console.WriteLine($"Повторная отправка уведомления #{NotificationId} (попыток: {retryCount}/{RetryLimit})...");
    }

    public virtual void SendNotification(string customHeader)
    {
        Console.WriteLine($"[{customHeader}] Отправка уведомления #{NotificationId}...");
    }

    void IArchivable.Archive()
    {
        IsArchived = true;
        Console.WriteLine($"[Явный архив] Уведомление #{NotificationId} успешно отправлено в архив.");
    }

    bool IArchivable.IsArchived
    {
        get => IsArchived;
        set => IsArchived = value;
    }
}

public class EmailNotification : Notification
{
    private string emailAddress;

    public string Subject { get; set; } = "Важное сообщение";
    public bool HasAttachment { get; set; } = false;
    public int SmtpPort { get; set; } = 587;

    public string EmailAddress
    {
        get => emailAddress;
        set => emailAddress = value;
    }

    public EmailNotification(int notificationId, string messageText, string emailAddress) 
        : base(notificationId, messageText, "Email")
    {
        EmailAddress = emailAddress;
    }

    public override void SendNotification()
    {
        IsSent = true;
        Console.WriteLine($"[Email] Отправка на {EmailAddress} (Порт: {SmtpPort}) | Тема: \"{Subject}\": \"{MessageText}\"");
    }

    public void AttachFile(string fileName)
    {
        HasAttachment = true;
        Console.WriteLine($"[Email] К письму #{NotificationId} прикреплен файл: {fileName}");
    }

    public void VerifyEmailDomain()
    {
        Console.WriteLine($"[Email] Домен адреса {EmailAddress} проверен и безопасен.");
    }
}

public class SmsNotification : Notification
{
    private string phoneNumber;

    public string MobileOperator { get; set; } = "Tele2";
    public int CharacterCount { get => MessageText.Length; }
    public bool IsFlashSms { get; set; } = false;

    public string PhoneNumber
    {
        get => phoneNumber;
        set => phoneNumber = value;
    }

    public SmsNotification(int notificationId, string messageText, string phoneNumber) 
        : base(notificationId, messageText, "SMS")
    {
        PhoneNumber = phoneNumber;
    }

    public override void SendNotification()
    {
        IsSent = true;
        Console.WriteLine($"[SMS] Отправка на номер {PhoneNumber} ({MobileOperator}) [Символов: {CharacterCount}]: \"{MessageText}\"");
    }

    public void FormatPhoneNumber()
    {
        Console.WriteLine($"[SMS] Номер телефона {PhoneNumber} отформатирован по международному стандарту.");
    }

    public void CheckDeliveryReport()
    {
        Console.WriteLine($"[SMS] Запрошен отчет о доставке SMS #{NotificationId}.");
    }
}

public class PushNotification : Notification
{
    public string DeviceToken { get; set; } = "token_xyz";
    public string AppVersion { get; set; } = "1.0.4";
    public bool SoundEnabled { get; set; } = true;

    public PushNotification(int notificationId, string messageText, string deviceToken) 
        : base(notificationId, messageText, "Push")
    {
        DeviceToken = deviceToken;
    }

    public override void SendNotification()
    {
        IsSent = true;
        Console.WriteLine($"[Push] Отправка на устройство (Token: {DeviceToken}, Звук: {SoundEnabled}): \"{MessageText}\"");
    }

    public void SendSilentPing() => Console.WriteLine($"[Push] Отправлен тихий пинг для токена {DeviceToken}.");
    public void UpdateBadgeCount(int count) => Console.WriteLine($"[Push] Бейдж приложения обновлен до значения: {count}.");
    public void RevokeToken() => Console.WriteLine($"[Push] Токен устройства {DeviceToken} аннулирован.");
}

public class SlackNotification : Notification
{
    public string ChannelName { get; set; } = "#general";
    public string BotName { get; set; } = "AlertBot";
    public bool MarkdownSupported { get; set; } = true;

    public SlackNotification(int notificationId, string messageText, string channelName) 
        : base(notificationId, messageText, "Slack")
    {
        ChannelName = channelName;
    }

    public override void SendNotification()
    {
        IsSent = true;
        Console.WriteLine($"[Slack] Канал {ChannelName} от бота {BotName}: \"{MessageText}\"");
    }

    public void AddReaction(string emoji) => Console.WriteLine($"[Slack] К сообщению #{NotificationId} добавлена реакция :{emoji}:");
    public void PinMessage() => Console.WriteLine($"[Slack] Сообщение #{NotificationId} закреплено в канале {ChannelName}.");
    public void ArchiveChannel() => Console.WriteLine($"[Slack] Канал {ChannelName} заархивирован системно.");
}

public class NotificationProcessor
{
    private readonly Notification _notification;

    public NotificationProcessor(Notification notification)
    {
        _notification = notification;
    }

    public void ProcessAndArchive()
    {
        Console.WriteLine($"--- Обработка через внедренную зависимость (ID: {_notification.NotificationId}) ---");
        _notification.Validate();
        _notification.SendNotification();
        IArchivable archivable = _notification;
        archivable.Archive();
        Console.WriteLine();
    }
}

public class NotificationManager<T> where T : Notification
{
    private List<T> notifications = new List<T>();

    public void Add(T notification)
    {
        notifications.Add(notification);
    }

    public void ProcessAll()
    {
        foreach (var item in notifications)
        {
            item.DisplayNotification();
            item.SendNotification();
            Console.WriteLine();
        }
    }
}


NotificationManager<Notification> manager = new NotificationManager<Notification>();

var email1 = new EmailNotification(1, "Привет!", "user@example.com");
email1.SenderName = "Администрация";
email1.Subject = "Добро пожаловать!";
email1.AttachFile("guide.pdf");

var sms1 = new SmsNotification(2, "Код подтверждения: 1234", "+79990000000");
sms1.MobileOperator = "MTS";
sms1.CheckDeliveryReport();

manager.Add(email1);
manager.Add(sms1);
manager.ProcessAll();

Console.WriteLine("=== Управление зависимостями (DI) ===");
Notification push = new PushNotification(3, "Новое системное обновление!", "device_token_999");
Notification slack = new SlackNotification(4, "Билд #42 успешно собран.", "#dev-ops");

NotificationProcessor processor1 = new NotificationProcessor(push);
processor1.ProcessAndArchive();

NotificationProcessor processor2 = new NotificationProcessor(slack);
processor2.ProcessAndArchive();

Console.WriteLine("=== Тест перегрузки методов ===");
Notification emailTest = new EmailNotification(5, "Тест перегрузки", "test@example.com");
emailTest.SendNotification(3);
emailTest.SendNotification("VIP-РАССЫЛКА");

Console.WriteLine("\n=== Тест явной реализации интерфейса ===");
IArchivable archivableSms = new SmsNotification(6, "Архивное SMS", "+78005553535");
archivableSms.Archive();
```
