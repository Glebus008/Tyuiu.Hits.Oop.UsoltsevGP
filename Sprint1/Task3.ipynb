<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Notification в C#, который будет представлять уведомления
пользователям. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Notification:
• Атрибуты: ID уведомления (NotificationId), Текст уведомления (MessageText),
Тип уведомления (Type).
• Методы:
o DisplayNotification(): метод для отображения уведомления
пользователю.
o SendNotification(): метод для отправки уведомления.
o GetNotificationDetails(): метод для получения деталей уведомления.
Требования к производным классам:
1. EmailУведомление (EmailNotification): Должно содержать дополнительные
атрибуты, такие как Адрес электронной почты (EmailAddress).
Метод SendNotification() должен быть переопределен для отправки
уведомления по электронной почте.
2. SMSУведомление (SMSNotification): Должно содержать дополнительные
атрибуты, такие как Номер телефона (PhoneNumber).
Метод SendNotification() должен быть переопределен для отправки
уведомления через SMS.
3. PushУведомление (PushNotification) (если требуется третий класс): Должно
содержать дополнительные атрибуты, такие как Платформа (Platform,
например, iOS или Android). Метод DisplayNotification() должен быть
переопределен для отображения уведомления на мобильной платформе.

Дополнительное задание

Добавьте к существующим классам конструктора классов в использованием сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// Базовый класс Notification с сеттерами и конструктором
class Notification
{
    private int notificationId;
    private string messageText;
    private string type;

    public int NotificationId
    {
        get => notificationId;
        set => notificationId = value;
    }

    public string MessageText
    {
        get => messageText;
        set => messageText = value;
    }

    public string Type
    {
        get => type;
        set => type = value;
    }

    public Notification(int notificationId, string messageText, string type)
    {
        NotificationId = notificationId;
        MessageText = messageText;
        Type = type;
    }

    public virtual void DisplayNotification()
    {
        Console.WriteLine($"[Уведомление #{NotificationId}] Тип: {Type} | Текст: \"{MessageText}\"");
    }

    public virtual void SendNotification()
    {
        Console.WriteLine($"Отправка стандартного уведомления #{NotificationId}...");
    }

    public virtual string GetNotificationDetails()
    {
        return $"ID: {NotificationId}, Тип: {Type}, Текст: {MessageText}";
    }
}

// Производный класс EmailNotification
class EmailNotification : Notification
{
    private string emailAddress;

    public string EmailAddress
    {
        get => emailAddress;
        set => emailAddress = value;
    }

    public EmailNotification(int notificationId, string messageText, string emailAddress) 
        : base(notificationId, messageText, "Email")
    {
        EmailAddress = emailAddress;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"[Email] Отправка письма на адрес {EmailAddress} (ID: {NotificationId}): \"{MessageText}\"");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Email: {EmailAddress}";
    }
}

// Производный класс SmsNotification
class SmsNotification : Notification
{
    private string phoneNumber;

    public string PhoneNumber
    {
        get => phoneNumber;
        set => phoneNumber = value;
    }

    public SmsNotification(int notificationId, string messageText, string phoneNumber) 
        : base(notificationId, messageText, "SMS")
    {
        PhoneNumber = phoneNumber;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"[SMS] Отправка сообщения на номер {PhoneNumber} (ID: {NotificationId}): \"{MessageText}\"");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Телефон: {PhoneNumber}";
    }
}

// Производный класс PushNotification
class PushNotification : Notification
{
    private string platform;

    public string Platform
    {
        get => platform;
        set => platform = value;
    }

    public PushNotification(int notificationId, string messageText, string platform) 
        : base(notificationId, messageText, "Push")
    {
        Platform = platform;
    }

    public override void DisplayNotification()
    {
        Console.WriteLine($"[Push | Платформа: {Platform}] Уведомление #{NotificationId}: \"{MessageText}\"");
    }

    public override string GetNotificationDetails()
    {
        return base.GetNotificationDetails() + $", Платформа: {Platform}";
    }
}

// Класс для демонстрации взаимодействия объектов между собой (Диспетчер уведомлений)
class NotificationDispatcher
{
    private string dispatcherName;

    public string DispatcherName
    {
        get => dispatcherName;
        set => dispatcherName = value;
    }

    public NotificationDispatcher(string dispatcherName)
    {
        DispatcherName = dispatcherName;
    }

    public void ProcessAndSend(Notification notification, string newText)
    {
        Console.WriteLine($"\n[Диспетчер {DispatcherName}] Обрабатывает уведомление #{notification.NotificationId}...");
        
        // Модифицируем текст через сеттер
        notification.MessageText = newText;
        
        notification.DisplayNotification();
        notification.SendNotification();
        
        Console.WriteLine($"[Диспетчер {DispatcherName}] Статус: Успешно отправлено.");
    }
}

// Исполняемый код (в самом верху ячейки)
var dispatcher = new NotificationDispatcher("Система-1");

Notification[] notifications = new[]
{
    (Notification)new EmailNotification(1, "Привет!", "user@example.com"),
    (Notification)new SmsNotification(2, "Временный код", "+7 (999) 000-00-00"),
    (Notification)new PushNotification(3, "Старый пуш", "iOS")
};

dispatcher.ProcessAndSend(notifications[0], "Ваш заказ успешно оформлен!");
dispatcher.ProcessAndSend(notifications[1], "Код подтверждения: 7749");
dispatcher.ProcessAndSend(notifications[2], "У вас новое сообщение от друга.");

Console.WriteLine("\n" + new string('=', 40) + "\nИтоговые детали всех объектов:");
foreach (var notif in notifications)
{
    Console.WriteLine(notif.GetNotificationDetails());
}
```
